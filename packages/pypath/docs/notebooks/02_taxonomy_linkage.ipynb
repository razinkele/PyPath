# Taxonomy Linkage in PyPath

## Overview
The taxonomy linkage system provides structured taxonomic classification for species groups. This notebook demonstrates taxonomy creation, validation, and integration with biodata.

```python
from pypath.core.params import TaxonLink, TaxonomyLinkage
```

## Part 1: Creating TaxonLink Records

The TaxonLink dataclass stores hierarchical taxonomic information:

```python
# Create a TaxonLink for a specific species
cod_taxon = TaxonLink(
    group_name="Cod",
    kingdom="Animalia",
    phylum="Chordata",
    class_name="Actinopterygii",
    order="Gadiformes",
    family="Gadidae",
    genus="Gadus",
    species="morhua",
    aphia_id=126417,
    common_name="Atlantic cod",
)

print("TaxonLink created:")
print(f"  Group: {cod_taxon.group_name}")
print(f"  Scientific name: {cod_taxon.genus} {cod_taxon.species}")
print(f"  Family: {cod_taxon.family}")
print(f"  WoRMS AphiaID: {cod_taxon.aphia_id}")
```

## Part 2: Creating a TaxonomyLinkage Container

Manage taxonomy for all groups in a model:

```python
# Create taxonomy container
taxonomy = TaxonomyLinkage()

# Add multiple species
herring_taxon = TaxonLink(
    group_name="Herring",
    kingdom="Animalia",
    phylum="Chordata",
    class_name="Actinopterygii",
    order="Clupeiformes",
    family="Clupeidae",
    genus="Clupea",
    species="harengus",
)

zooplankton_taxon = TaxonLink(
    group_name="Copepods",
    kingdom="Animalia",
    phylum="Arthropoda",
    class_name="Copepoda",
)

taxonomy.add_taxon(cod_taxon)
taxonomy.add_taxon(herring_taxon)
taxonomy.add_taxon(zooplankton_taxon)

print(f"Added {len(taxonomy)} taxa to container")
print(f"Groups: {taxonomy.group_names()}")
```

## Part 3: Accessing Taxonomic Data

```python
# Get specific taxon
cod_retrieved = taxonomy.get_taxon("Cod")
print("Retrieved taxon for Cod:")
print(f"  Scientific name: {cod_retrieved.genus} {cod_retrieved.species}")
print(f"  Family: {cod_retrieved.family}")

# Check existence
print(f"\nContains Cod: {taxonomy.has_taxon('Cod')}")
print(f"Contains Tuna: {taxonomy.has_taxon('Tuna')}")
```

## Part 4: Completeness Validation

Verify taxonomy completeness and hierarchy:

```python
# Validate and check completeness
is_valid = taxonomy.validate()
print(f"Taxonomy is valid: {is_valid}")

# Completeness for each group
stats = taxonomy.completeness_stats()
print("\nCompleteness statistics:")
for group, completeness in stats.items():
    print(f"  {group}: {completeness:.1%}")
```

## Part 5: DataFrame Serialization

```python
# Convert to DataFrame
taxonomy_df = taxonomy.to_dataframe()
print("Taxonomy DataFrame:")
print(taxonomy_df)

# Save to CSV
taxonomy_df.to_csv("taxonomy_backup.csv", index=False)
print("\nSaved to taxonomy_backup.csv")
```

## Part 6: Real-World Example - Baltic Species

Complete workflow for Baltic Sea model initialization:

```python
# Create Baltic taxonomy
baltic_taxonomy = TaxonomyLinkage()

# Define key Baltic species
baltic_species = [
    ("Cod", {"genus": "Gadus", "species": "morhua"}),
    ("Herring", {"genus": "Clupea", "species": "harengus"}),
    ("Sprat", {"genus": "Sprattus", "species": "sprattus"}),
    ("Flounder", {"genus": "Platichthys", "species": "flesus"}),
]

for group, info in baltic_species:
    taxon = TaxonLink(
        group_name=group,
        kingdom="Animalia",
        phylum="Chordata",
        class_name="Actinopterygii",
        genus=info["genus"],
        species=info["species"],
    )
    baltic_taxonomy.add_taxon(taxon)

print(f"Added {len(baltic_taxonomy)} Baltic species")
export_df = baltic_taxonomy.to_dataframe()[["group_name", "genus", "species"]]
print(export_df.to_string(index=False))
```

## Performance Tips

- TaxonomyLinkage lookup is O(1) for group names
- DataFrame conversion is O(n) where n = number of species
- Cache completeness results to avoid repeated calculations
- Use batch operations when loading multiple taxa
